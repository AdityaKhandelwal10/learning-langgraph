```python
from typing import Dict, TypedDict, List
import math
from langgraph.graph import StateGraph
```

```python
class AgentState(TypedDict):
    values : List[int]
    name : str
    result : str
    operation : str
```

```python
# def sum(state : AgentState) -> AgentState:
#     """
#     This node takes in a list of values and returns the sum of the values
#     """
#     state["result"] = "Hi " + state["name"] + " the sum of your values is " + str(sum(state["values"]))
#     return state

# def multiply(state : AgentState) -> AgentState:
#     """
#     This node takes in a list of values and returns the product of the values
#     """
#     state["result"] = "Hi " + state["name"] + " the product of your values is " + str(math.prod(state["values"]))
#     return state
    
def check_operation(state : AgentState) -> AgentState:
    """
    This node takes in a list of values and returns the product of the values
    """
    if state["operation"] == "sum":
        state["result"] = "Hi " + state["name"] + " the sum of your values is " + str(sum(state["values"]))
        return state
    elif state["operation"] == "multiply":
        state["result"] = "Hi " + state["name"] + " the product of your values is " + str(math.prod(state["values"]))
        return state
    else:
        raise ValueError("Invalid operation")
    return state
```

```python
# graph = StateGraph(AgentState)
# graph.add_node("sum", sum_node)
# graph.set_entry_point("sum")
# graph.set_finish_point("sum")

# graph.compile()

graph = StateGraph(AgentState)

graph.add_node("check_operation", check_operation)

graph.set_entry_point("check_operation")
graph.set_finish_point("check_operation")
```

```python
app = graph.compile()

result = app.invoke({"values" : [1,2,3,4,5], "name" : "Aditya", "operation" : "sum"})
```
